# Monte Carlo Convergence and Accuracy Analysis

This notebook studies the convergence and accuracy of Monte Carlo methods for option pricing.

Monte Carlo option pricing estimates the expected discounted payoff by simulating a large number of possible future paths of the underlying asset.

Since the estimate is based on random samples, the calculated option price can vary depending on the number of simulated paths.

In this notebook, we investigate how the number of Monte Carlo simulations affects:

- The estimated option price
- The accuracy of the estimate
- The standard error of the estimator
- The convergence of the Monte Carlo method

A European call option is used as the primary example.
