# Advanced Variance Reduction in Monte Carlo Option Pricing

This notebook explores advanced variance reduction techniques used in Monte Carlo simulation for option pricing.

The objective of variance reduction is to obtain more accurate estimates using the same number of simulated paths.

The primary techniques explored in this notebook are:

- Control Variates
- Importance Sampling

The performance of these methods will be compared with standard Monte Carlo simulation.

## 1. Control Variates

Control variates reduce the variance of a Monte Carlo estimator by using another random variable that is correlated with the quantity of interest and whose expected value is known.

Suppose we wish to estimate

$$
E[X].
$$

Let $Y$ be a related random variable such that

$$
E[Y] = \mu_Y
$$

is known.

The control variate estimator is

$$
\hat{\mu}_{CV}
=
\bar{X}
+
c(\mu_Y-\bar{Y}),
$$

where $c$ is the control variate coefficient.

For option pricing, we use the discounted option payoff as $X$ and the discounted terminal asset price as the control variate $Y$.
